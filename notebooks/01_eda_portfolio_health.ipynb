# Day 1 Learning Log: Dataset Audit & Missing Value Inspection

### Key Commands
* `df.shape`: Returns dimensions as `(rows, columns)`.
* `df.isnull().sum()`: Counts total missing values per feature.
* `low_memory=False`: Prevents Pandas warnings when loading large CSV files with mixed data types per column.

### Credit Analytics Insights
1. **Missing Data carries signal:** High null percentages in credit columns (e.g., `mths_since_last_delinq`) do not necessarily mean corrupt data—they often indicate that a borrower has never defaulted or been delinquent.
2. **Text Cleaning required:** Rates like `int_rate`, `emp_length`, `earliest_cr_line`, and durations like `term` (`" 36 months"`) are loaded as strings and must be cleaned before casting to float/int.

# Cleanning numeric data strings

```python
#creating a copy of the original dataset to clean and transform
df_clean = df_loan_dataset.copy()
```

```python
# 1. Clean Interest Rate (e.g., " 13.99%" -> 13.99)
if 'int_rate' in df_clean.columns and df_clean['int_rate'].dtype == 'object':
    df_clean['int_rate_clean'] = df_clean['int_rate'].astype(str).str.replace('%', '').str.strip()
    df_clean['int_rate_clean'] = pd.to_numeric(df_clean['int_rate_clean'], errors='coerce')

# 2. Clean Term (e.g., " 36 months" -> 36)
if 'term' in df_clean.columns and df_clean['term'].dtype == 'object':
    df_clean['term_months'] = df_clean['term'].astype(str).str.extract(r'(\d+)').astype(float)

# 3. Clean Employment Length ("10+ years" -> 10.0, "< 1 year" -> 0.5)
def clean_emp_length(val):
    if pd.isna(val) or str(val).lower() == 'nan':
        return np.nan
    val_str = str(val)
    if '10+' in val_str:
        return 10.0
    elif '< 1' in val_str:
        return 0.5
    else:
        extracted = ''.join(filter(str.isdigit, val_str))
        return float(extracted) if extracted else np.nan

if 'emp_length' in df_clean.columns:
    df_clean['emp_length_years'] = df_clean['emp_length'].apply(clean_emp_length)

print("'int_rate', 'term', and 'emp_length' columns cleaned successfully.")
```

```python
# 4. Handle Delinquency Signal & High-Null Risk Features
delinq_cols = ['delinq_2yrs', 'mths_since_last_delinq', 'pub_rec', 'mths_since_last_record']
for col in delinq_cols:
    if col in df_clean.columns:
        df_clean[f'{col}_is_null'] = df_clean[col].isnull().astype(int)
        df_clean[f'{col}_filled'] = df_clean[col].fillna(0)

# 5. Clean Key Ratios
if 'annual_inc' in df_clean.columns:
    df_clean['annual_inc_clean'] = df_clean['annual_inc'].fillna(df_clean['annual_inc'].median())

# 6. Export cleaned version
df_clean.to_csv('../data/processed/cleaned_loans.csv', index=False)
print("Data cleaning & feature extraction complete!")
print(f"Saved {df_clean.shape[0]:,} rows and {df_clean.shape[1]} columns to ../data/processed/cleaned_loans.csv")
```
