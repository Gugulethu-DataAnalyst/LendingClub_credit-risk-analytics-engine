```python
#Null value summary dataframe
null_summary = pd.DataFrame({
    'data_types': df_loan_dataset.dtypes,
    'null_count': df_loan_dataset.isnull().sum(),
    'null_percentage': (df_loan_dataset.isnull().sum() / df_loan_dataset.shape[0]) * 100
})

#Filtering to show only the columns with missing values
missing_col = null_summary[null_summary['null_count'] > 0].sort_values(by='null_percentage', ascending=False)

print(f'Total columns missing values: {missing_col.shape[0]}')
missing_col.head(10)
```

```python
#Checking the raw formats for key text fields
cols_to_check = ['int_rate', 'term', 'emp_length', 'revol_util', 'earliest_cr_line']
for cols in cols_to_check:
    print(f"---SAMPLE VALUES FOR {cols}---")
    print(df_loan_dataset[cols].dropna().head(50).tolist())
    print()
```

# Day 1 Learning Log: Dataset Audit & Missing Value Inspection

### Key Commands
* `df.shape`: Returns dimensions as `(rows, columns)`.
* `df.isnull().sum()`: Counts total missing values per feature.
* `low_memory=False`: Prevents Pandas warnings when loading large CSV files with mixed data types per column.

### Credit Analytics Insights
1. **Missing Data carries signal:** High null percentages in credit columns (e.g., `mths_since_last_delinq`) do not necessarily mean corrupt data—they often indicate that a borrower has never defaulted or been delinquent.
2. **Text Cleaning required:** Rates like `int_rate`, `emp_length`, `earliest_cr_line`, and durations like `term` (`" 36 months"`) are loaded as strings and must be cleaned before casting to float/int.
